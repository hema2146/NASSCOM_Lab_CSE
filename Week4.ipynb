```python
import sympy as sp
def g(x):
    return x**3 + 2*x
x = sp.symbols('x')
expr = x**3 + 2*x
# 1. Numerical derivative at x = 2 using h = 1e-6
h = 1e-6
slope_at_2 = (g(2 + h) - g(2)) / h
print(
    "Numerical g'(2):",
    round(slope_at_2, 4),
    "(exact = 14)"
)
# 2. Symbolic derivative
deriv = sp.diff(expr, x)

print("d/dx (x**3 + 2*x) =", deriv)

print(
    "Symbolic g'(2) =",
    deriv.subs(x, 2)
)


# 3. Evaluate symbolic derivative at x = 2 and compare
symbolic_value = deriv.subs(x, 2)

print("\nComparison:")
print("Numerical derivative:", slope_at_2)
print("Symbolic derivative:", symbolic_value)

print(
    "Are they approximately equal?",
    abs(slope_at_2 - float(symbolic_value)) < 1e-5
)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2


h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
# 1. Forward pass: z1, h = ReLU(z1), y_hat, loss
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1     # input -> hidden
Wb = np.random.randn(8, 1) * 0.1
z1 = Xb @ Wa
h = np.maximum(0, z1)

y_hat = h @ Wb

# Mean squared error loss
loss = np.mean((y_hat - Yb) ** 2)

print("Loss before:", loss)


# 2. Backward pass: dy, dWb, dh, dz1, dWa
n = Xb.shape[0]

# dL/dy_hat
dy = (2 / n) * (y_hat - Yb)

# dL/dWb
dWb = h.T @ dy

# dL/dh
dh = dy @ Wb.T

# ReLU derivative
dz1 = dh * (z1 > 0)

# dL/dWa
dWa = Xb.T @ dz1


# 3. One step with lr = 0.05; print loss before and after
lr = 0.05

Wa -= lr * dWa
Wb -= lr * dWb

# Forward pass again to calculate new loss
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb

loss_after = np.mean((y_hat - Yb) ** 2)

print("Loss after:", loss_after)
```

```python
xv = 0.0          # starting point
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)        # the derivative
    xv = xv - lr * grad        # step against the gradient
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
# 1. Hessian of x**4 + y**2
import sympy as sp

x, y = sp.symbols('x y')

f = x**4 + y**2
H = sp.hessian(f, (x, y))

print('Hessian:')
sp.pprint(H)


# 2. Gradient descent to minimise (x - 7)**2
xv = 0.0
lr = 0.1

for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad

print('Final x:', xv)
```
