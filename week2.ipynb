```python
import datetime
import random

DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}
raw_sensor_data = []

start_time = datetime.datetime.now().replace(microsecond=0)

for i in range(100):
    timestamp = start_time + datetime.timedelta(minutes=i)
    device_id = f"DEV{i + 1:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_value, max_value = READING_RANGES[device_type]
    value = round(random.uniform(min_value, max_value), 2)

    raw_sensor_data.append((
        timestamp.strftime('%Y-%m-%d %H:%M:%S'),
        device_id,
        device_type,
        location,
        value
    ))

device_stats = {}
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (
        float('inf'),
        float('-inf')
    )
for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device_type]

    new_min = min(current_min, value)
    new_max = max(current_max, value)

    device_stats[device_type] = (
        new_min,
        new_max
    )

print("Device statistics:\n")

for device_type, stats in device_stats.items():
    print(
        f"{device_type}: "
        f"Minimum = {stats[0]}, "
        f"Maximum = {stats[1]}"
    )
```

```python
def calculate_average_temperature(temp_data_list):

    if not temp_data_list:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading['temperature_celsius']

    average = total / len(temp_data_list)

    return average


temperature_readings = [
    {
        'timestamp': '2026-08-17 10:00:00',
        'device_id': 'DEV001',
        'location': 'Lab1',
        'temperature_celsius': 25.5
    },
    {
        'timestamp': '2026-08-17 10:01:00',
        'device_id': 'DEV002',
        'location': 'Lab2',
        'temperature_celsius': 27.5
    },
    {
        'timestamp': '2026-08-17 10:02:00',
        'device_id': 'DEV003',
        'location': 'OfficeA',
        'temperature_celsius': 24.0
    }
]

average_temperature = calculate_average_temperature(
    temperature_readings
)

print(f"Average temperature: {average_temperature:.2f}°C")
```
