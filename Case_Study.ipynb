```python
battery_data = [[0,32,400,20,95],
                [1,35,395,22,88],
                [2,38,390,25,80],
                [3,40,385,28,70],
                [4,42,380,30,60],
                [5,45,375,32,48]]
for reading in battery_data:
  print(f"Time: {reading[0]}h, "
  f"Temperature: {reading[1]}C, "
  f"Voltage: {reading[2]}V, "
  f"Current: {reading[3]}A, "
  f"Charge: {reading[4]}%")
```

```python
for reading in battery_data:
  temperature = reading[1]
  if temperature > 40:
    print(f"Time {reading[0]}h : Hight Temperature Alert")
  else:
    print(f"Time {reading[0]}h : Temperature Normal")
```

```python
for reading in battery_data:
  charge = reading[4]
  if charge < 50:
    print(f"Time {reading[0]}h : Low Battery Alert")
  else:
    print(f"Time {reading[0]}h : Battery Charge Normal")
```

```python
def battery_status(charge):
  if charge>70:
    return "High"
  elif 50<=charge<=70:
    return "Medium"
  else:
    return "Low"
for reading in battery_data:
  print(f"Time {reading[0]}h : "
  f"Charge = {reading[4]}%, "
  f"Status = {battery_status(reading[4])}")
```
