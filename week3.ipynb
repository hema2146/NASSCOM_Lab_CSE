```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

print('A symmetric?', np.allclose(A, A.T))

theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])

# 1. Scaling matrix (x * 2, y * 0.5)
S = np.array([[2, 0],
              [0, 0.5]])

# 2. Rotation matrix (45 degrees)
t = np.radians(45)

R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

# 3. Apply scaling and rotation
scaled_tri = S @ tri
rotated_scaled_tri = R @ scaled_tri

# Plot original triangle
plt.plot(
    np.append(tri[0], tri[0, 0]),
    np.append(tri[1], tri[1, 0]),
    'bo-',
    label='Original'
)

# Plot transformed triangle
plt.plot(
    np.append(rotated_scaled_tri[0], rotated_scaled_tri[0, 0]),
    np.append(rotated_scaled_tri[1], rotated_scaled_tri[1, 0]),
    'ro-',
    label='Scaled + Rotated'
)

# Add coordinate axes
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Original vs Transformed Triangle')
plt.legend()
plt.grid(True)
plt.axis('equal')

plt.show()
```

```python
A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np

C = np.array([[4., 1.],
              [2., 3.]])

# 1. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

# 2. Print eigenvalues
print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

# 3. Verify A v = lambda v for the first eigenvector
v = eigenvectors[:, 0]
lambda_ = eigenvalues[0]

left_side = C @ v
right_side = lambda_ * v

print("\nFirst eigenvector:")
print(v)

print("\nC @ v:")
print(left_side)

print("\nlambda * v:")
print(right_side)

print("\nVerification:")
print(np.allclose(left_side, right_side))
```

```python
import numpy as np

A2 = np.array([[3., 2.],
               [1., 4.]])
b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])

# 1. Solve A2 x = b2
x = np.linalg.solve(A2, b2)

print("Solution x:")
print(x)

# 2. Rank of A2
rank_A2 = np.linalg.matrix_rank(A2)

print("\nRank of A2:")
print(rank_A2)

# 3. Cosine similarity function
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

cos_cat_dog = cosine_similarity(cat, dog)
cos_cat_car = cosine_similarity(cat, car)

print("\nCosine similarity (cat, dog):")
print(cos_cat_dog)

print("\nCosine similarity (cat, car):")
print(cos_cat_car)

# Determine which is more similar
if cos_cat_dog > cos_cat_car:
    print("\nCat is more similar to Dog.")
else:
    print("\nCat is more similar to Car.")
```
