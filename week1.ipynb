Functions

```python
def greet():
  print("Hello! Welcome to Python Functions.")
greet()
```

```python
def greet(name):
  print(f"Hello!, {name}")
greet("Honey")
greet("Akhila")
```

```python
def greet_user(name="Guest"):
  print(f"Welcome, {name}!")
greet_user()
greet_user("Seetha")
```

```python
def print_names(*names):
  print("Names passed:",names)
  for name in names:
    print("Hello",name)
print_names("Honey","Anitha","Banu")
```

```python
def print_info(**info):
  print("User Info:")
  for key,value in info.items():
    print(f"{key} = {value}")
print_info(name = "Honey",age = 21,city = "Hyderabad")
```

```python
def show_details(*args,**kwargs):
  print("Args:",args)
  print("Kwargs:",kwargs)
show_details(1,2,3,name = "Honey",age = 21)
```

```python
def multiply(a,b):
  return a*b
def square(n):
  return multiply(n,n)
print("Square of 4:",square(4))
```

Anonymous Functions

```python
from functools import reduce
numbers = [1,2,3]
a = lambda x : x**2
res = list(map(lambda y : y**2,numbers))
b = list(filter(lambda x : x%2 == 0,res))
total = reduce(lambda x,y : x+y , res)
print(a(3))
print(res)
print(b)
print(total)
```

Lists

```python
fruits = ['Banana','Papaya','Mango','Pineapple']
print("Original List:",fruits)
print("First Fruit:",fruits[0])
print("Last Fruit:",fruits[-1])
fruits.append('Jackfruit')
print("After appending:",fruits)
fruits.remove('Papaya')
print("After removing Papaya:",fruits)
print("Number of fruits:",len(fruits))
```

Sets

```python
unique_numbers={1,2,3,4,1}
print("Set:",unique_numbers)
unique_numbers.add(16)
print("After adding 16:",unique_numbers)
unique_numbers.remove(1)
print("After removing 1:",unique_numbers)
```

```python
set_a = {1,2,3}
set_b = {3,4,5}
print("Union:",set_a | set_b)
print("Intersection:",set_a & set_b)
print("Difference(A-B):",set_a - set_b)
print("Difference(B-A):",set_b - set_a)
```

Dictionaries

```python
student = {
    'name' : 'Honey',
    'age' : 21,
    'course' : 'python'
}
print("Dictionary:",student)
print("Student Name:",student['name'])
student['grade'] = 'A'
print("After adding grade:",student)
del student['age']
print("After deleting age:",student)
for key,value in student.items():
  print(f"{key} : {value}")
```

```python
keys = ['id','name','marks']
values = [101,'Honey',98]
student_info = dict(zip(keys,values))
print("Students info from zip:",student_info)
```

Lists : Indexing & Slicing

```python
my_list = [10,30,49,35,28,64]
print("First element:",my_list[0])
print("Last element:",my_list[-1])
```

```python
print("Elements from index 1 to 3:",my_list[1:4])
print("From start to index 2:",my_list[:3])
print("From index 2 to end:",my_list[2:])
print("Every second element:",my_list[::2])
print("Reverse list:",my_list[::-1])
```

Tuples: Indexing & Slicing

```python
my_tuple = ('a','b','e','r','p','s','h')
print("Element at index 0:",my_tuple[0])
print("Element at index -2:",my_tuple[-2])
print("Slice from index 1 to 3:",my_tuple[1:4])
print("Tuple reverse:",my_tuple[::-1])
```

Strings: Indexing & Slicing

```python
text = "hello world"
print("Character at index 0:",text[0])
print("Last Character:",text[-1])
print("First 6 characters:",text[:6])
print("From 6 to end:",text[6:])
print("Every 2nd character:",text[::2])
print("Reversed string:",text[::-1])
```
