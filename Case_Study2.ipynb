```python
def F(x):
    return 40 + 5*x - 0.2*x**2

def dF(x):
    return 5 - 0.4*x

x_critical = 5 / 0.4

print(f"Derivative dF/dx: 5 - 0.4*x")
print(f"Value of x where dF(x) = 0: {x_critical} hours\n")

print("x (Hours) | F(x) (Score) | dF(x) (Rate of Change)")
print("-" * 45)
for x in range(1, 16):
    print(f"{x:<10} | {F(x):<12.2f} | {dF(x):<.2f}")

print("\nInterpretation:")
print(f"At {x_critical} study hours, the rate of score change (dF/dx) is 0,")
print(f"yielding the maximum score of {F(x_critical):.2f}. Studying beyond {x_critical} hours")
print("causes dF/dx to turn negative, leading to lower predicted scores due to burnout.")
```
